Создать базовый класс PaymentMethod в C#, который будет представлять различные способы оплаты. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;


// Базовый класс PaymentMethod
public class PaymentMethod
{
    public int PaymentMethodId { get; set; }
    public string MethodName { get; set; }
    public decimal MinAmount { get; set; }


    // Конструктор
    public PaymentMethod(int paymentMethodId, string methodName, decimal minAmount)
    {
        PaymentMethodId = paymentMethodId;
        MethodName = methodName;
        MinAmount = minAmount;
    }


    // для обработки платежа
    public virtual void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Обработка платежа на сумму {amount} руб.");
    }


    // для проверки минимальной суммы
    public virtual bool CheckMinimumAmount(decimal amount)
    {
        if (amount >= MinAmount)
        {
            Console.WriteLine($"Сумма {amount} руб. соответствует минимальной сумме платежа");
            return true;
        }
        else
        {
            Console.WriteLine($"Сумма {amount} руб. меньше минимальной суммы платежа ({MinAmount} руб.)");
            return false;
        }
    }


    // для получения деталей способа оплаты
    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"ID: {PaymentMethodId}, Название: {MethodName}, Мин. сумма: {MinAmount} руб.");
    }
}


// Онлайн оплата
public class OnlinePayment : PaymentMethod
{
    public string PaymentUrl { get; set; }


    public OnlinePayment(int paymentMethodId, string methodName, decimal minAmount, string paymentUrl)
        : base(paymentMethodId, methodName, minAmount)
    {
        PaymentUrl = paymentUrl;
    }


    // Переопределяем метод обработки платежа
    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Онлайн платеж на сумму {amount} руб. через систему: {PaymentUrl}");
        Console.WriteLine("Перенаправление на страницу оплаты...");
        Console.WriteLine("Онлайн платеж успешно завершен!");
    }


    // Переопределяем метод получения деталей
    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"URL платежной системы: {PaymentUrl}");
    }
}


// Банковский перевод
public class BankTransfer : PaymentMethod
{
    public string BankData { get; set; }


    public BankTransfer(int paymentMethodId, string methodName, decimal minAmount, string bankData)
        : base(paymentMethodId, methodName, minAmount)
    {
        BankData = bankData;
    }


    // Переопределяем проверку минимальной суммы с учетом комиссий
    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * 0.01m; // 1% комиссия
        decimal totalAmount = amount + commission;
        
        Console.WriteLine($"Проверка банковского перевода:");
        Console.WriteLine($"Сумма: {amount} руб.");
        Console.WriteLine($"Комиссия (1%): {commission} руб.");
        Console.WriteLine($"Итого: {totalAmount} руб.");
        Console.WriteLine($"Требуемый минимум: {MinAmount} руб.");
        
        if (totalAmount >= MinAmount)
        {
            Console.WriteLine("Сумма с учетом комиссий соответствует требованиям");
            return true;
        }
        else
        {
            Console.WriteLine("Ошибка: итоговая сумма с комиссиями меньше минимальной!");
            return false;
        }
    }


    // Переопределяем метод получения деталей
    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Банковские данные: {BankData}");
    }
}


// Наличные
public class CashPayment : PaymentMethod
{
    public string CashPickupPoint { get; set; }


    public CashPayment(int paymentMethodId, string methodName, decimal minAmount, string cashPickupPoint)
        : base(paymentMethodId, methodName, minAmount)
    {
        CashPickupPoint = cashPickupPoint;
    }


    // Переопределяем метод получения деталей
    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Место выдачи наличных: {CashPickupPoint}");
    }


    // Переопределяем метод обработки платежа
    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Оплата наличными на сумму {amount} руб.");
        Console.WriteLine($"Наличные можно внести в пункте: {CashPickupPoint}");
        Console.WriteLine("Оплата наличными зарегистрирована!");
    }
}


// Классы уведомлений

public class Notification
{
    public int NotificationId { get; set; }
    public string MessageText { get; set; }
    public string Type { get; set; }

    // Конструктор
    public Notification(int notificationId, string messageText, string type)
    {
        NotificationId = notificationId;
        MessageText = messageText;
        Type = type;
    }

    // для отображения уведомления
    public virtual void DisplayNotification()
    {
        Console.WriteLine($"Уведомление (ID: {NotificationId}, Тип: {Type}) - {MessageText}");
    }

    // для отправки уведомления
    public virtual void SendNotification()
    {
        Console.WriteLine("Отправляем уведомление...");
    }

    // для получения деталей уведомления
    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationId}, Тип: {Type}, Сообщение: {MessageText}";
    }
}

public class EmailNotification : Notification
{
    public string EmailAddress { get; set; }

    public EmailNotification(int notificationId, string messageText, string emailAddress)
        : base(notificationId, messageText, "Email")
    {
        EmailAddress = emailAddress;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправляем на почту {EmailAddress}: {MessageText}");
        Console.WriteLine("Email уведомление успешно отправлено!");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Почта: {EmailAddress}";
    }
}

public class SMSNotification : Notification
{
    public string PhoneNumber { get; set; }

    public SMSNotification(int notificationId, string messageText, string phoneNumber)
        : base(notificationId, messageText, "SMS")
    {
        PhoneNumber = phoneNumber;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправляем смс на {PhoneNumber}: {MessageText}");
        Console.WriteLine("SMS уведомление успешно отправлено!");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Телефон: {PhoneNumber}";
    }
}

public class PushNotification : Notification
{
    public string Platform { get; set; }

    public PushNotification(int notificationId, string messageText, string platform)
        : base(notificationId, messageText, "Push")
    {
        Platform = platform;
    }

    public override void DisplayNotification()
    {
        Console.WriteLine($"[{Platform}] Пуш уведомление: {MessageText}");
        Console.WriteLine($"ID уведомления: {NotificationId}");
    }

    public override void SendNotification()
    {
        Console.WriteLine($"Отправляем пуш уведомление на {Platform} устройство: {MessageText}");
        Console.WriteLine("Пуш уведомление успешно отправлено!");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Платформа: {Platform}";
    }
}


class Program
{
    static void Main()
    {
        Console.WriteLine("=== СИСТЕМА ОПЛАТЫ ===");
        Console.WriteLine("Создаем способы оплаты...\n");


        // Создаем экземпляры платежей
        OnlinePayment online = new OnlinePayment(1, "Онлайн оплата", 10, "https://pay.example.com");
        BankTransfer bank = new BankTransfer(2, "Банковский перевод", 100, "Сбербанк, р/с 1234567890");
        CashPayment cash = new CashPayment(3, "Наличные", 1, "ул. Центральная, д. 15");


        Console.WriteLine("1. ОНЛАЙН ОПЛАТА");
        Console.WriteLine("-------------------");
        online.GetPaymentDetails();
        online.CheckMinimumAmount(50);
        online.ProcessPayment(50);
        Console.WriteLine();


        Console.WriteLine("2. БАНКОВСКИЙ ПЕРЕВОД");
        Console.WriteLine("-------------------");
        bank.GetPaymentDetails();
        bank.CheckMinimumAmount(80);
        bank.ProcessPayment(80);
        Console.WriteLine();


        Console.WriteLine("3. НАЛИЧНЫЕ");
        Console.WriteLine("-------------------");
        cash.GetPaymentDetails();
        cash.CheckMinimumAmount(500);
        cash.ProcessPayment(500);
        Console.WriteLine();


        Console.WriteLine("=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");
        Console.WriteLine("Обработка через массив базового класса:\n");


        PaymentMethod[] payments = { online, bank, cash };
        decimal[] amounts = { 150, 200, 300 };


        for (int i = 0; i < payments.Length; i++)
        {
            Console.WriteLine($"Платеж #{i + 1} через {payments[i].MethodName}:");
            payments[i].GetPaymentDetails();
            payments[i].CheckMinimumAmount(amounts[i]);
            payments[i].ProcessPayment(amounts[i]);
            Console.WriteLine();
        }

        // Создаем уведомления для каждого способа оплаты
        Console.WriteLine("=== УВЕДОМЛЕНИЯ ===\n");

        EmailNotification emailNotification = new EmailNotification(1, "Платеж успешно выполнен через онлайн систему", "user@example.com");
        SMSNotification smsNotification = new SMSNotification(2, "Платеж выполнен через банковский перевод", "+7-900-123-45-67");
        PushNotification pushNotification = new PushNotification(3, "Платеж наличными зарегистрирован", "Android");

        emailNotification.SendNotification();
        Console.WriteLine(emailNotification.GetNotificationDetails() + "\n");

        smsNotification.SendNotification();
        Console.WriteLine(smsNotification.GetNotificationDetails() + "\n");

        pushNotification.DisplayNotification();
        Console.WriteLine(pushNotification.GetNotificationDetails() + "\n");

        Console.WriteLine("Программа завершена!");
        Console.WriteLine("Нажмите любую клавишу для выхода...");
        Console.ReadKey();
    }
}
```
