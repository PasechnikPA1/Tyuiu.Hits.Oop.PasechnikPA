----

Создать базовый класс PaymentMethod в C#, который будет представлять различные способы оплаты. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

class PaymentMethod
{
    public int PaymentMethodId;
    public string MethodName;
    public decimal MinAmount;
    
    public int ID
    {
        get { return PaymentMethodId; }
        set { PaymentMethodId = value; }
    }
    
    public string Name
    {
        get { return MethodName; }
        set { MethodName = value; }
    }
    
    public decimal MinimumAmount
    {
        get { return MinAmount; }
        set { MinAmount = value; }
    }
    
    public virtual void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Обработка платежа на сумму {amount} руб.");
    }
    
    public virtual bool CheckMinimumAmount(decimal amount)
    {
        if (amount >= MinAmount)
        {
            Console.WriteLine($"Сумма {amount} руб. соответствует минимальной сумме платежа ({MinAmount} руб.)");
            return true;
        }
        else
        {
            Console.WriteLine($"Сумма {amount} руб. меньше минимальной суммы платежа ({MinAmount} руб.)");
            return false;
        }
    }
    
    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"ID: {PaymentMethodId}, Название: {MethodName}, Мин. сумма: {MinAmount} руб.");
    }
}

class OnlinePayment : PaymentMethod
{
    public string PaymentUrl;
    
    public string Url
    {
        get { return PaymentUrl; }
        set { PaymentUrl = value; }
    }
    
    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Онлайн платеж на сумму {amount} руб. через систему: {PaymentUrl}");
        Console.WriteLine("Перенаправление на страницу оплаты...");
    }
    
    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"URL платежной системы: {PaymentUrl}");
    }
}

class BankTransfer : PaymentMethod
{
    public string BankData;
    
    public string Bank
    {
        get { return BankData; }
        set { BankData = value; }
    }
    
    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * 0.01m; 
        decimal totalAmount = amount + commission;
        
        if (totalAmount >= MinAmount)
        {
            Console.WriteLine($"Сумма {amount} руб. + комиссия {commission} руб. = {totalAmount} руб.");
            Console.WriteLine($"Соответствует минимальной сумме платежа ({MinAmount} руб.)");
            return true;
        }
        else
        {
            Console.WriteLine($"Сумма {amount} руб. + комиссия {commission} руб. = {totalAmount} руб.");
            Console.WriteLine($"Меньше минимальной суммы платежа ({MinAmount} руб.)");
            return false;
        }
    }
    
    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Банковские данные: {BankData}");
    }
}

class CashPayment : PaymentMethod
{
    public string CashPickupPoint;
    
    public string PickupPoint
    {
        get { return CashPickupPoint; }
        set { CashPickupPoint = value; }
    }
    
    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Место выдачи наличных: {CashPickupPoint}");
    }
    
    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Оплата наличными на сумму {amount} руб.");
        Console.WriteLine($"Наличные можно внести в пункте: {CashPickupPoint}");
    }
}


Console.WriteLine("Демонстрация работы с PaymentMethod \n");

OnlinePayment online = new OnlinePayment();
online.PaymentMethodId = 1;
online.MethodName = "Онлайн оплата";
online.MinAmount = 10;
online.PaymentUrl = "https://pay.example.com";

BankTransfer bank = new BankTransfer();
bank.PaymentMethodId = 2;
bank.MethodName = "Банковский перевод";
bank.MinAmount = 50;
bank.BankData = "АО 'Пример Банк', р/с 1234567890";

CashPayment cash = new CashPayment();
cash.PaymentMethodId = 3;
cash.MethodName = "Наличные";
cash.MinAmount = 1;
cash.CashPickupPoint = "ул. Примерная, д. 123";

Console.WriteLine("1. Онлайн оплата:");
online.GetPaymentDetails();
online.CheckMinimumAmount(5);
online.CheckMinimumAmount(15);
online.ProcessPayment(100);
Console.WriteLine();

Console.WriteLine("2. Банковский перевод:");
bank.GetPaymentDetails();
bank.CheckMinimumAmount(45);
bank.CheckMinimumAmount(60);
bank.ProcessPayment(200);
Console.WriteLine();

Console.WriteLine("3. Наличные:");
cash.GetPaymentDetails();
cash.CheckMinimumAmount(0.5m);
cash.CheckMinimumAmount(5);
cash.ProcessPayment(50);
```
