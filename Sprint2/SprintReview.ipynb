```polyglot-notebook

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    List<string> GetTeamMembers();
}

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public float Salary { get; set; }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"имя: {Name}, должность: {Position}, зарплата: {Salary}");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public override void PrintInfo()
    {
        Console.WriteLine($"имя: {Name}, должность: {Position}, зарплата: {Salary}, отдел: {Department}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    public List<string> TeamMembers { get; set; } = new List<string>();

    public override void PrintInfo()
    {
        Console.WriteLine($"имя: {Name} должность: {Position}, зарплата: {Salary}, отдел: {Department}, проект: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"отчет по проекту: {ProjectName}. количество участников: {TeamMembers.Count}";
    }

    public List<string> GetTeamMembers()
    {
        return TeamMembers;
    }
}


Employee emp = new Employee { Name = "Иван", Position = "разработчик", Salary = 70000 };
Manager mgr = new Manager { Name = "Андрей", Position = "менеджер", Salary = 90000, Department = "ИТ" };
ProjectManager pm = new ProjectManager
{
    Name = "Светлана",
    Position = "проектный менеджер",
    Salary = 120000,
    Department = "IT",
    ProjectName = "автоматизация",
    TeamMembers = new List<string> { "Иван", "Андрей", "Светлана" }
};

emp.PrintInfo();
mgr.PrintInfo();
pm.PrintInfo();
Console.WriteLine(pm.GenerateReport());
Console.WriteLine("Состав команды: " + string.Join(", ", pm.GetTeamMembers()));
```
