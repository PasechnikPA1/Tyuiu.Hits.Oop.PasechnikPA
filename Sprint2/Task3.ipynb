<h2 style="color:DodgerBlue">Описание проекта:</h2>

Создать базовый класс PaymentMethod в C#, который будет представлять различные способы оплаты. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```

using System;
using System.Collections.Generic;
using System.Threading.Tasks;


interface IOnlinePaymentFeatures
{
    string GetSecurityToken();
    void SaveTransactionHistory(decimal amount);
}

interface IBankPaymentFeatures
{
    string GenerateReceipt(decimal amount);
    void NotifyBank(decimal amount);
}


class PaymentMethod
{
    private int paymentMethodId;
    private string methodName;
    private decimal minAmount;
    private decimal maxAmount;
    private bool isActive;
    private string category;

    public PaymentMethod(int id, string name, decimal minAmount, decimal maxAmount, 
                        bool isActive, string category)
    {
        ID = id;
        Name = name;
        MinimumAmount = minAmount;
        MaximumAmount = maxAmount;
        IsActive = isActive;
        Category = category;
    }

    public int ID { get => paymentMethodId; set => paymentMethodId = value; }
    public string Name { get => methodName; set => methodName = value; }
    public decimal MinimumAmount { get => minAmount; set => minAmount = value; }
    public decimal MaximumAmount { get => maxAmount; set => maxAmount = value; }
    public bool IsActive { get => isActive; set => isActive = value; }
    public string Category { get => category; set => category = value; }

    public virtual void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Обработка платежа на сумму {amount} руб.");
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        if (amount >= minAmount && amount <= maxAmount)
        {
            Console.WriteLine($"Сумма {amount} руб. находится в допустимом диапазоне от {minAmount} до {maxAmount} руб.");
            return true;
        }
        Console.WriteLine($"Сумма {amount} руб. находится вне допустимого диапазона от {minAmount} до {maxAmount} руб.");
        return false;
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"ID: {paymentMethodId}, Название: {methodName}, Мин. сумма: {minAmount} руб., Макс. сумма: {maxAmount} руб.");
        Console.WriteLine($"Статус: {(isActive ? "Активен" : "Неактивен")}, Категория: {category}");
    }
}


class SimplePayment : PaymentMethod
{
    private string description;

    public SimplePayment(int id, string name, decimal minAmount, decimal maxAmount, 
                        bool isActive, string category, string description) 
        : base(id, name, minAmount, maxAmount, isActive, category)
    {
        Description = description;
    }

    public string Description { get => description; set => description = value; }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Описание: {description}");
    }
}


class BankTransfer : PaymentMethod
{
    private string bankData;
    private decimal commissionRate;

    public BankTransfer(int id, string name, decimal minAmount, decimal maxAmount, 
                        bool isActive, string category, string bankData, decimal commissionRate) 
        : base(id, name, minAmount, maxAmount, isActive, category)
    {
        Bank = bankData;
        CommissionRate = commissionRate;
    }

    public string Bank { get => bankData; set => bankData = value; }
    public decimal CommissionRate { get => commissionRate; set => commissionRate = value; }

    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * commissionRate;
        decimal totalAmount = amount + commission;
        if (totalAmount >= MinimumAmount && totalAmount <= MaximumAmount)
        {
            Console.WriteLine($"Сумма {amount} руб. + комиссия {commission} руб. = {totalAmount} руб.");
            Console.WriteLine($"Соответствует допустимому диапазону от {MinimumAmount} до {MaximumAmount} руб.");
            return true;
        }
        Console.WriteLine($"Сумма {amount} руб. + комиссия {commission} руб. = {totalAmount} руб.");
        Console.WriteLine($"Не соответствует допустимому диапазону от {MinimumAmount} до {MaximumAmount} руб.");
        return false;
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Банковские данные: {bankData}, Комиссия: {commissionRate * 100}%");
    }
}


class OnlineBankTransfer : BankTransfer, IOnlinePaymentFeatures, IBankPaymentFeatures
{
    private string securityToken;

    public OnlineBankTransfer(int id, string name, decimal minAmount, decimal maxAmount,
                            bool isActive, string category, string bankData, decimal commissionRate)
        : base(id, name, minAmount, maxAmount, isActive, category, bankData, commissionRate)
    {
        SecurityToken = Guid.NewGuid().ToString();
    }

    public string SecurityToken { get => securityToken; set => securityToken = value; }

    public string GetSecurityToken()
    {
        return $"Токен безопасности: {securityToken}";
    }

    public void SaveTransactionHistory(decimal amount)
    {
        Console.WriteLine($"Сохранена история транзакции на сумму {amount} руб.");
    }

    public string GenerateReceipt(decimal amount)
    {
        return $"Квитанция №{Guid.NewGuid()} на сумму {amount} руб.";
    }

    public void NotifyBank(decimal amount)
    {
        Console.WriteLine($"Уведомление банку о платеже на сумму {amount} руб.");
    }

    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine("Обработка онлайн-банковского перевода:");
        Console.WriteLine(GetSecurityToken());
        Console.WriteLine(GenerateReceipt(amount));
        NotifyBank(amount);
        base.ProcessPayment(amount);
        SaveTransactionHistory(amount);
    }
}


var online = new OnlineBankTransfer(1, "Онлайн-банковский перевод", 50, 50000, true, "Онлайн", 
    "АО 'Пример Банк', р/с 1234567890", 0.01m);

var bank = new BankTransfer(2, "Банковский перевод", 50, 50000, true, "Банковский",
    "АО 'Пример Банк', р/с 1234567890", 0.01m);

var simple = new SimplePayment(3, "Простой платёж", 1, 10000, true, "Базовый",
    "Стандартный способ оплаты");


Console.WriteLine(" Демонстрация работы с PaymentMethod \n");


Console.WriteLine(" Доступные способы оплаты \n");
online.GetPaymentDetails();
Console.WriteLine();
bank.GetPaymentDetails();
Console.WriteLine();
simple.GetPaymentDetails();


Console.WriteLine("\n Обработка платежей \n");


Console.WriteLine("1. Онлайн-банковский перевод:");
online.GetPaymentDetails();
online.CheckMinimumAmount(45);
online.ProcessPayment(100);


Console.WriteLine("\n2. Банковский перевод:");
bank.GetPaymentDetails();
bank.CheckMinimumAmount(45);
bank.ProcessPayment(200);


Console.WriteLine("\n3. Простой платёж:");
simple.GetPaymentDetails();
simple.CheckMinimumAmount(5);
simple.ProcessPayment(50);
```
