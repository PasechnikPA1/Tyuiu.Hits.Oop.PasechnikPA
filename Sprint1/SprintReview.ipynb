```polyglot-notebook
using System;
using System.Collections.Generic;

public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"{name} - {price} руб. (остаток: {quantity})";
    }
}

public class Seller
{
    private string name;
    private string employedId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> products;

    public Seller(string name, string employedId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employedId = employedId;
        this.salary = salary;
        this.contactInfo = contactInfo;
        products = new List<Product>();
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployedId
    {
        get { return employedId; }
        set { employedId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"Продано: {quantity} шт. {product.Name}");
        }
        else
        {
            Console.WriteLine($"Недостаточно: {product.Name}");
        }
    }

    public string GetSellerInfo()
    {
        return $"{name} (ID: {employedId})";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers;
    private List<Product> products;

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
        sellers = new List<Seller>();
        products = new List<Product>();
    }

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void ListProducts()
    {
        Console.WriteLine("\nТовары в магазине:");
        foreach (var product in products)
        {
            Console.WriteLine($"- {product.GetProductInfo()}");
        }
    }

    public string GetStoreInfo()
    {
        return $"{storeName} ({location})";
    }
}

Product laptop = new Product("Ноутбук", 50000, 10, "Электроника", "Игровой ноутбук");


Seller seller1 = new Seller("Иван Иванов","1", 40000, "ivan@store.com");


Store store = new Store("ТехноМир", "ул. Центральная, 123", "9:00-21:00");


store.AddProduct(laptop);

store.AddSeller(seller1);

seller1.AddProduct(laptop);



Console.WriteLine($"Магазин: {store.GetStoreInfo()}");
Console.WriteLine($"Продавец: {seller1.GetSellerInfo()}");
Console.WriteLine($"Товар: {laptop.GetProductInfo()}");
Console.WriteLine($"Общая стоимость: {laptop.GetTotalPrice()} руб.");

seller1.SellProduct(laptop, 2);


phone.UpdateQuantity(5);
Console.WriteLine($"\nОбновленный остаток ноутбуков: {laptop.Quantity}");


store.ListProducts();
```
