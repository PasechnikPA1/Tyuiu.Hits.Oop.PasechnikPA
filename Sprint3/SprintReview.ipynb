```polyglot-notebook
using System;
using System.Collections.Generic;
using System.Linq;


abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"{Name}, цена: {Price} руб.";
    }

    public override string ToString() => GetInfo();

    public static bool operator ==(Component c1, Component c2)
    {
        if (ReferenceEquals(c1, c2)) return true;
        if (ReferenceEquals(c1, null) || ReferenceEquals(c2, null)) return false;
        return c1.Name == c2.Name && c1.Price == c2.Price;
    }

    public static bool operator !=(Component c1, Component c2)
    {
        return !(c1 == c2);
    }

    public override bool Equals(object obj)
    {
        var other = obj as Component;
        return this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

interface IInstallable
{
    void Install();
}

class CPU : Component, IInstallable
{
    public CPU(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"CPU: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Установлен процессор {Name}.");
}

class RAM : Component, IInstallable
{
    public RAM(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"RAM: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Установлена память {Name}.");
}

class GPU : Component, IInstallable
{
    public GPU(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"GPU: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Установлена видеокарта {Name}.");
}

class HDD : Component, IInstallable
{
    public HDD(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"HDD: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Установлен накопитель {Name}.");
}

class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удалён из компьютера.");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден.");
        }
    }

    public decimal GetTotalPrice()
    {
        return components.Sum(c => c.Price);
    }

    public void ShowComponents()
    {
        Console.WriteLine("Состав компьютера:");
        foreach (var comp in components)
        {
            Console.WriteLine(comp);
        }
    }
}

static class ComputerFactory
{
    public static Computer CreateBasicComputer()
    {
        var computer = new Computer();
        computer.AddComponent(new CPU("Intel i5", 15000));
        computer.AddComponent(new RAM("Kingston 8GB", 5000));
        computer.AddComponent(new HDD("Toshiba 1TB", 4000));
        computer.AddComponent(new GPU("NVIDIA RTX 3060", 30000));
        return computer;
    }

    public static Computer CreateGamingComputer()
    {
        var computer = new Computer();
        computer.AddComponent(new CPU("Intel i9", 40000));
        computer.AddComponent(new RAM("Corsair 32GB", 15000));
        computer.AddComponent(new GPU("NVIDIA RTX 4080", 80000));
        computer.AddComponent(new HDD("Samsung SSD 2TB", 12000));
        return computer;
    }
}

// --- Демонстрация (выполняйте как отдельную ячейку) ---

var basicPC = ComputerFactory.CreateBasicComputer();
basicPC.ShowComponents();
Console.WriteLine($"Общая стоимость: {basicPC.GetTotalPrice()} руб.\n");

var gamingPC = ComputerFactory.CreateGamingComputer();
gamingPC.ShowComponents();
Console.WriteLine($"Общая стоимость: {gamingPC.GetTotalPrice()} руб.\n");

// Удаление компонента
var toRemove = new RAM("Kingston 8GB", 5000);
basicPC.RemoveComponent(toRemove);
basicPC.ShowComponents();
Console.WriteLine($"Общая стоимость после удаления: {basicPC.GetTotalPrice()} руб., но компьютер не работает");
```
